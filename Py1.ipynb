```python
#Count Function
total_tickets = df_ke["No."].count()
pro_tickets = df_ke[df_ke["Request Type.1"]=="Proactive"]["No."].count()
re_tickets = df_ke[df_ke["Request Type.1"]=="Reactive"]["No."].count()
ke_closed = df_ke[df_ke["Status"] == "Closed"]["No."].count()
    
total_tickets, pro_tickets, re_tickets, ke_closed
```
